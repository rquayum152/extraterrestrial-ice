## Imports and Constants ##

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import scipy.constants
```

## Plotting Ice Habitat Temperature as a Function of Time (Figure 7a) ##

```python
# Load data
file_path_1 = "your file location here/gsr_pt_4_full_seasonality_results_post_power.xlsx" # results of temperature of ground and inner ice base as a function of time from 40% GSR with power added to the ground transient solution
file_path_2 = "your file location here/gsr_pt_100_full_seasonality_results_post_power.xlsx" # results of temperature of ground and inner ice base as a function of time from 100% GSR transient solution

sheet1 = "temp_powered_inner_ice" # 40% GSR inner ice temp. vs. time
sheet2 = "temp_powered_ground" # 40% GSR ground temp vs. time
sheet3 = "temp_inner_ice" # 100% GSR inner ice temp vs. time
sheet4 = "temp_ground" # 100% GSR ground temp vs. time

inner_ice_df_40 = pd.read_excel(file_path_1, sheet_name=sheet1)
ground_df_40 = pd.read_excel(file_path_1, sheet_name=sheet2)

inner_ice_df_100 = pd.read_excel(file_path_2, sheet_name=sheet3)
ground_df_100 = pd.read_excel(file_path_2, sheet_name=sheet4)


fig, ax = plt.subplots(figsize=(5, 4))

# 40% GSR powered inner ice
ax.plot(
    inner_ice_df_40['time'] / 86400, # time
    inner_ice_df_40['avg'], # average temperature
    linewidth=2,
    label="40% GSR Powered Inner Ice"
)

ax.fill_between(
    inner_ice_df_40['time'] / 86400,
    inner_ice_df_40['min'], # minimum temperature
    inner_ice_df_40['max'], # maximum temperature
    alpha=0.15
)

# 40% GSR powered ground
ax.plot(
    ground_df_40['time'] / 86400,
    ground_df_40['avg'],
    linewidth=2,
    label="40% GSR Powered Ground"
)

ax.fill_between(
    ground_df_40['time'] / 86400,
    ground_df_40['min'],
    ground_df_40['max'],
    alpha=0.15
)

# 100% GSR inner ice
ax.plot(
    inner_ice_df_100['time'] / 86400,
    inner_ice_df_100['avg'],
    linewidth=2,
    label="100% GSR Inner Ice"
)

ax.fill_between(
    inner_ice_df_100['time'] / 86400,
    inner_ice_df_100['min'],
    inner_ice_df_100['max'],
    alpha=0.15
)

# 100% GSR ground
ax.plot(
    ground_df_100['time'] / 86400,
    ground_df_100['avg'],
    linewidth=2,
    label="100% GSR Ground"
)

ax.fill_between(
    ground_df_100['time'] / 86400,
    ground_df_100['min'],
    ground_df_100['max'],
    alpha=0.15
)

# Labels and formatting
ax.set_xlabel("Time (days)")
ax.set_ylabel("Temperature (°C)")

ax.legend(
    loc='upper center',
    bbox_to_anchor=(0.5, -0.18), ncol=2)
ax.grid(True, which="both", ls="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Plotting Power Entering the Habitat Ground vs. Time (Figure 7b) ##

```python
file_path_2 = "your file location here/power_added_reg_ground_40.xlsx" # power added to the system to heat the dome ground to habitable temperatures in 40% GSR simulation

power_df = pd.read_excel(file_path_2)

fig, ax = plt.subplots(figsize=(5, 4))

ax.plot(
    power_df['time']/86400, # seconds to days
    power_df['power_added']/1000,
    linewidth=2,
    label="Power Added to Ground"
)

ax.set_ylabel("Power (kW)")
ax.set_xlabel("Time (days)")

ax.grid(True, which="both", ls="--", alpha=0.4)
plt.tight_layout()
plt.show()
```
