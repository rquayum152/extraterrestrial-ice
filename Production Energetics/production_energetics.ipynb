## Imports ##

```python
import numpy as np
from scipy.interpolate import interp1d
```

## Energy of Melting Ice Calculation (Equation 3) ##

```python
# Ice temperature data (K)
T_data = np.array([
    173.15, 183.15, 193.15, 203.15, 213.15,
    223.15, 233.15, 238.15, 243.15, 248.15,
    253.15, 258.15, 263.15, 268.15, 273.15
])

# Specific heat capacity of ice data (kJ kg^-1 K^-1) from https://www.engineeringtoolbox.com
cp_data = np.array([
    1.389, 1.463, 1.536, 1.609, 1.681,
    1.751, 1.818, 1.851, 1.882, 1.913,
    1.943, 1.972, 2.000, 2.027, 2.050
])

# Interpolate specific heat as a function of temperature
cp_ice_temp_dependent = interp1d(T_data, cp_data, kind='linear')

def sensible_heat_ice(T0, T):
    # Create temperature array between initial and final temperatures
    T_array = np.linspace(T0, T, 1000)

    # Calculate cp at each temperature
    cp_array = cp_ice_temp_dependent(T_array)

    # Integrate cp(T) over temperature
    epsilon = np.trapezoid(cp_array, T_array)

    return epsilon # kJ/kg

# Calculate energy required to heat ice from 200 K to 273 K
epsilon = sensible_heat_ice(200, 273)

print('Sensible heating energy =', round(epsilon, 1), 'kJ/kg')
```

## Ice Generation Calculation ##

```python
processing_energy = 600 # kJ/kg
power_source = 100 # kJ/s
rho_ice = 917 # ice density, kg/m^3
habitat_thickness = 0.5 # m

production_rate = (power_source/processing_energy)*86400 # converted to kg of ice/day
print('Ice Production Rate =', round(production_rate/1000, 1), 'tons/day') # converted to tons of ice/day

area_day = production_rate/rho_ice # m^2 of ice generated/day
habitat_day = area_day/habitat_thickness # area of ice habitat/day
habitat_year = habitat_day*365 # area of ice habitat/year
print('Ice Habitat Area Generated per Year =', round(habitat_year/10000, 2), 'hectares/year') # converted to hectares of ice/year
```
